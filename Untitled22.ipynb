```python
from PyPDF2 import PdfReader

reader = PdfReader("/content/2605.05578v1.pdf")
print(f"Number of pages: {len(reader.pages)}")

# Extract text from all pages
parts = []
for page in reader.pages:
    text = page.extract_text()
    if text:
        parts.append(text)

text_body = "\n".join(parts)
print(text_body[:1000])  # Print first 1000 chars for preview
```

```python
from transformers import AutoTokenizer

model_ckpt = "facebook/bart-large-cnn"
tokenizer = AutoTokenizer.from_pretrained(model_ckpt)

def count_tokens(text):
    return len(tokenizer.encode(text, truncation=False))

num_tokens = count_tokens(text_body)
print(f"Number of tokens in extracted text: {num_tokens}")
```

```python
import os
from langchain_community.chat_models import ChatHuggingFace
from langchain_core.prompts import ChatPromptTemplate

# os.environ["HUGGINGFACEHUB_API_TOKEN"] = "your_huggingface_token"

llm = ChatHuggingFace(
    repo_id="HuggingFaceH4/zephyr-7b-beta",
    temperature=0.2,
    max_new_tokens=512,
)

prompt = ChatPromptTemplate.from_messages([
    ("system", "You are a helpful AI assistant. Summarize the following research paper in clear markdown format."),
    ("human", "{paper_content}")
])

chain = prompt | llm

output = chain.invoke({"paper_content": text_body})
print(output.content)
```

```python
with open("Summary.txt", "w", encoding="utf-8") as f:
    f.write(output.content)

print("Summary saved to Summary.txt")
```
